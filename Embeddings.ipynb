```python
import numpy as np
import pandas as pd
```

```python
class EMBED():
    def __init__(self, vocabSize :int=10000, dim :int=100, _ngrams :int=2):
        self._ndim = dim
        self.vocabSize = vocabSize
        self.word2Vec = {"<UNK>": 0}
        self._ngrams = _ngrams

    def _loadVocab(self, text :list[str]):
        ctr=1
        for sentence in text:
            for word in sentence.split():
                if not word in self.word2Vec.keys():
                    self.word2Vec[word] = ctr
                    ctr+=1

                if ctr >= self.vocabSize:
                    return

        if ctr<self.vocabSize:
            self.vocabSize = ctr
            print(f"Not enough words.Setting Vocab Size to {ctr}")
            
    def _loadCoccurence(self, text :list[str]):
        for sentence in text:
            sentence = sentence.split(" ")
            for i in range(len(sentence)):
                wordVec = 0
                if sentence[i] in self.word2Vec.keys():
                    wordVec = self.word2Vec[sentence[i]]
                
                gram = [None]*(self._ngrams*2)
                ctr = i-self._ngrams
                for j in range(len(gram)):
                    if ctr==i:
                        ctr+=1
                    if ctr>=0 and ctr<len(sentence):
                        if not sentence[ctr] in self.word2Vec.keys():
                            gram[j] = 0
                        else:
                            gram[j] = self.word2Vec[sentence[ctr]]

                        #gram matrix is loaded
                        self.coccurence_matrix[wordVec, gram[j]] = self.coccurence_matrix[wordVec][gram[j]] + 1
        
    def _genEmbedding(self):
        embeddings = self.word2Vec.copy()
        for key, value in embeddings:
            embeddings[key] = np.random.rand((self._ndim))
        
    def __train__(self, text :list[str]):
        self._loadVocab(text)
        self.coccurence_matrix = np.zeros(shape=(self.vocabSize, self.vocabSize), dtype=np.int16)
        self._loadCoccurence(text)
        #self._genEmbedding()
```

```python
embedding = EMBED(vocabSize = 5000)
```
